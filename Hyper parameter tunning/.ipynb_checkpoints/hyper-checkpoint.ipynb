```python
from sklearn import svm, datasets
import pandas as pd
```

```python
iris = datasets.load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
dir(iris)
```

```python
df['flower'] = iris.target
```

```python
df['flower'] = df['flower'].apply(lambda x: iris.target_names[x])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3)
```

```python
model = svm.SVC(kernel='rbf', C=30, gamma='auto')
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cv1 = cross_val_score(svm.SVC(kernel='rbf', C=30, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
cv2 = cross_val_score(svm.SVC(kernel='linear', C=30, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
cv3 = cross_val_score(svm.SVC(kernel='rbf', C=20, gamma='auto'), iris.data, iris.target, cv=5) # cv means cross validation
```

```python
import numpy as np
```

```python
np.average(cv1)
```

```python
np.average(cv2)
```

```python
np.average(cv3)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
clf = GridSearchCV(svm.SVC(gamma='auto'), {
    'C': [ 1, 10, 20],
    'kernel': ['rbf', 'linear']
}, cv=5, return_train_score=False)

clf.fit(iris.data, iris.target)
clf.cv_results_
```

```python
df = pd.DataFrame(clf.cv_results_)
df
```

```python
df[['param_C', 'param_kernel', 'mean_test_score']]
```
