```python
# load iris dataset and measure the accuracy of different algorithm to figureo ut which is better
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
dir(iris)
```

```python
cross_val_score(LogisticRegression(max_iter=1000), iris.data, iris.target)
```

```python
cross_val_score(LinearRegression(), iris.data, iris.target)
```

```python
cross_val_score(RandomForestClassifier(), iris.data, iris.target)
```

```python
cross_val_score(SVC(), iris.data, iris.target)
```

```python
X = iris.data
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
def get_score(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return accuracy_score(y_test, y_pred)
```

```python
get_score(LinearRegression(), X_train, X_test, y_train, y_test)
```

```python
get_score(LogisticRegression(), X_train, X_test, y_train, y_test)
```

```python
get_score(RandomForestClassifier(), X_train, X_test, y_train, y_test)
```

```python
get_score(SVC(), X_train, X_test, y_train, y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
folds = StratifiedKFold(n_splits=3)

scaler = StandardScaler()
```

```python
logistic = []
linear = []
sv = []
rf = []

for train_idx, test_idx in folds.split(X, y):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.fit_transform(X_test)

    print("SVM:", get_score(SVC(), X_train, X_test, y_train, y_test))
    print("Logistic:", get_score(LogisticRegression(max_iter=1000), X_train, X_test, y_train, y_test))
    print("RF:", get_score(RandomForestClassifier(), X_train, X_test, y_train, y_test))
```
