```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model

df = pd.read_csv('dataset.csv')
df
```

```python
%matplotlib inline
plt.scatter(df.price, df['size'], marker='+', color='green')
plt.title('Scatter plot')
plt.ylabel('Size in sq ft')
plt.xlabel('Price')
plt.show()
```

```python
reg = linear_model.LinearRegression()
# size will be input and we predict is price
reg.fit(df[['size']], df[['price']])
```

```python
prediction = reg.predict([[20]])
print('predicted price is', prediction[0])
```

```python
# price will be input and we predict is size
reg.fit(df[['price']], df[['size']])
predict = reg.predict([[100]])
print('Predicted size', predict[0])
```
